```python
import json
import shapely
import shapely.geometry
import pyproj
import geopandas as gp
```

```python
with open('resources/merged_waters_fixed_with_fence_buffer.geojson') as f:
    d = json.load(f)
```

```python
# read in water holes

ct = {}
points = []

for feat in d['features']:
    if feat['geometry']['type'] != 'MultiPoint':
        continue
        
    
        
    t = feat['properties']['TYPE']
    if t in ct:
        ct[t] += 1
    else:
        ct[t] = 1
```

```python
wgs84 = pyproj.CRS('EPSG:4326')


def create_rectangle(lat, lon, w = 1000):
    """
    Create circle with with/height around given POINT.

    See: https://gis.stackexchange.com/questions/310872/create-polygon-with-length-width-and-a-center-coordinate
    """    
    gf = gp.GeoDataFrame({'lat':lat, 'lon':lon, 'width': w, 'height': w}, index=[0])
    gf.crs = {'init' :'epsg:4326'}

    gf['center'] = gf.apply(lambda x: shapely.geometry.Point(x['lon'], x['lat']), axis=1)
    gf = gf.set_geometry('center')

    gf = gf.to_crs(epsg = 3857)

    gf['center'] = gf['center'].buffer(1)
    gf['polygon'] = gf.apply(lambda x: shapely.affinity.scale(x['center'], x['width'], x['height']), axis=1)
    gf = gf.set_geometry('polygon')
    
    gf = gf.to_crs(epsg = 4326)
    #gf.plot()
    
    return (gf['polygon'][0])
    

    
    
```

```python
p = create_rectangle(53.55191637937063, 10.080030145969282)
```
