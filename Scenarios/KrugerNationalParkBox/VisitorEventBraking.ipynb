```python
import numpy as np

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.pyplot import cm

from shapely.geometry import MultiPoint
from shapely.geometry import Point, LineString, Polygon
import shapely.wkt
```

```python
df = pd.read_csv('./bin/Debug/net5.0/Tourist.csv', sep=";", parse_dates=['DateTime'])
ids = list(df['ID'].unique())

df['CarVelocity'] = df['CarVelocityInt']
```

```python
df['CarVelocity_kmh'] = df['CarVelocity'] * 3.6
```

```python
def state_to_int(state):
    if state == 'Driving':
        return 1
    if state == 'Braking':
        return 2
    if state == 'Looking':
        return 3
    
    return 0
```

```python
dfx = df[df['ID'] == 'f0efb595-1b4d-4b98-8121-980ba4a28ed6']
dfx = dfx.set_index('DateTime')
```

```python
dfx['CarVelocity'].plot()
plt.title('Velocity of agent (m/s)')
plt.show()
```

```python
dfx['StateInt'] = dfx['State'].apply(state_to_int)
dfx['StateInt'].plot()
plt.title('State: 1, 2, 3')
plt.show()
```
